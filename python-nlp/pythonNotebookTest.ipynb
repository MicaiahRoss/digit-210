```python
import os

season1_tokens = []
season5_tokens = []


for filename in os.listdir("Project Text Files/01"):
    if filename.endswith(".txt"):
        with open(os.path.join("Project Text Files/01", filename), 'r', encoding='utf-8') as file:
            text = file.read()
            tokens = text.split()
            lowerTokens = [w.lower() for w in tokens]
            season1_tokens.extend(lowerTokens)


for filename in os.listdir("Project Text Files/05"):
    if filename.endswith(".txt"):
        with open(os.path.join("Project Text Files/05", filename), 'r', encoding='utf-8') as file:
            text = file.read()
            tokens = text.split()
            lowerTokens = [w.lower() for w in tokens]
            season5_tokens.extend(lowerTokens)


season1Text = nltk.Text(season1_tokens)
season5Text = nltk.Text(season5_tokens)


season1_diversity = len(set(season1_tokens)) / len(season1_tokens)
season5_diversity = len(set(season5_tokens)) / len(season5_tokens)

print(f"Season 1 Lexical Diversity: {season1_diversity:.4f}")
print(f"Season 5 Lexical Diversity: {season5_diversity:.4f}")
```

```python
from nltk.tokenize import word_tokenize
from nltk.tag import pos_tag

season1_tagged = pos_tag(season1_tokens)
season5_tagged = pos_tag(season5_tokens)

tagsIwant = ['JJ']

season1_adjectives = [word for word, tag in season1_tagged if tag in tagsIwant]
season5_adjectives = [word for word, tag in season5_tagged if tag in tagsIwant]

season1_adjSet = set(season1_adjectives)
season5_adjSet = set(season5_adjectives)

print(f"Season 1 adjectives sample: {list(season1_adjSet)[:20]}")
print(f"Season 5 adjectives sample: {list(season5_adjSet)[:20]}")
print(season1_adjSet)
print(season5_adjSet)
```

6. Develop the code in the previous block by copying it into the next code block, and here:
   * Develop code to find out about the top 15 most ambiguous words in your adjective or verb token list, and
   * Write your output to a neatly formatted text file for reading. The text output file should provide a clear heading to provide a title for each corpus you're comparing. *(5 pts)*

```python
from nltk.corpus import wordnet as wn

season1_ambiguity = []
for w in sorted(season1_adjSet):
    lemma = wn.morphy(w)
    lemma = lemma if lemma else w
    synsets = wn.synsets(lemma)
    if synsets:
        season1_ambiguity.append((w, len(synsets)))

season1_top15 = sorted(season1_ambiguity, key=lambda x: x[1], reverse=True)[:15]

season5_ambiguity = []
for w in sorted(season5_adjSet):
    lemma = wn.morphy(w)
    lemma = lemma if lemma else w
    synsets = wn.synsets(lemma)
    if synsets:
        season5_ambiguity.append((w, len(synsets)))

season5_top15 = sorted(season5_ambiguity, key=lambda x: x[1], reverse=True)[:15]

print("=== Season 1 - Top 15 Most Ambiguous Adjectives ===")
for word, count in season1_top15:
    print(f"Word: {word} | Number of Synsets (Ambiguity): {count}")

print("\n=== Season 5 - Top 15 Most Ambiguous Adjectives ===")
for word, count in season5_top15:
    print(f"Word: {word} | Number of Synsets (Ambiguity): {count}")

# write the output file
with open("ambiguity_output.txt", "w", encoding="utf-8") as f:
    f.write("=== Season 1 - Top 15 Most Ambiguous Adjectives ===\n")
    for word, count in season1_top15:
        f.write(f"Word: {word} | Number of Synsets (Ambiguity): {count}\n")
    
    f.write("\n=== Season 5 - Top 15 Most Ambiguous Adjectives ===\n")
    for word, count in season5_top15:
        f.write(f"Word: {word} | Number of Synsets (Ambiguity): {count}\n")

print("Output written to ambiguity_output.txt")
```
