```python
%matplotlib inline
import nltk
import matplotlib.pyplot as plt
from nltk import word_tokenize, FreqDist


with open("avatarSpeeches.txt", "r", encoding="utf-8") as file:
    text = file.read()
    tokens = word_tokenize(text) 
    lower_tokens = [w.lower() for w in tokens if w.isalpha()] 
    avatar_text = nltk.Text(lower_tokens)


diversity = len(set(avatar_text)) / len(avatar_text)
print(f"Lexical Diversity: {diversity}")


fdist = FreqDist(avatar_text)
fdist.plot(20)
```

```python
from nltk.corpus import stopwords


stop_words = set(stopwords.words('english'))


filtered_tokens = [w for w in lower_tokens if w not in stop_words]
filtered_text = nltk.Text(filtered_tokens)


fdist_filtered = FreqDist(filtered_text)
fdist_filtered.plot(20)
```
